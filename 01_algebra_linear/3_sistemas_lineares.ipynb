# 2.0 Determinantes:

Para calcular o determinante de uma matriz, basta usar a função linalg.det(A):

```python
# Criando a matriz m1:

m1 = np.array([
    [1,5,9],
    [7,-3,4],
    [0,2,3]
])

print(m1)
```

```python
# Calculando o determinante, temos:

detm1 = round(np.linalg.det(m1), 2)
detm1
```

# 3.0 Sistemas Lineares:

Para resolver um Sistema Linear, poderíamos escrever na forma matricial e solucionar com as ferramentas que vimos utilizando matriz inversa, assim:

$$
\begin{aligned}
A X &= B \\
A^{-1} A X &= A^{-1} B \\
I_n X &= A^{-1} B \\
X &= A^{-1} B
\end{aligned}
$$


O que daria muito trabalho, para facilitar, podemos apenas usar o método solve(). Vamos criar um Sistema Linear e resolver com o solve.

> **Observação:** Embora teoricamente possamos escrever  
> $$
> X=A^{-1}B
> $$
> na prática computacional normalmente evitamos
> calcular a inversa explicitamente. Para resolver sistemas,
> `np.linalg.solve(A, B)` é geralmente a opção preferida.

## 3.1 Com a função solve():

```python
# Criando a Matriz A:

A = np.array([
    [1,2,5,7,-1],
    [1,1,0,5,6],
    [1,0,6,7,9],
    [-1,0.3,0,-2,1],
    [5,6,0,0,1]
])

# Criando a matriz B:

B = np.array([
    [4],
    [3],
    [0],
    [9],
    [-1]
])
```

```python
# Resolvendo o sistema com o solve:

X_s = np.linalg.solve(A,B)
X_s
```

## 3.2 Com matriz:

```python
# Primeiro, temos que ver se o det A não é zero, assim:

detA = np.linalg.det(A)
detA
```

```python
# Como não é zero, podemos calcaular a sua inversa:

invA = np.linalg.inv(A)
invA
```

```python
# Para obter a Matriz X da solução, basta multiplicar A^-1 por B:

X_m = invA@B
X_m
```

```python
# Para fazer uma verificação e comparar arrays, poderíamos fazer:
np.array_equal(X_s, X_m)
```

```python
# porém, como temos muitos números decimais, seria mais fácil certeiro assim:
np.allclose(X_s, X_m)
```
