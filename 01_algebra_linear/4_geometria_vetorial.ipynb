# 2.0 Produto Interno

Dados dois vetores

$$
u=(u_1,u_2,\ldots,u_n)
$$

e

$$
v=(v_1,v_2,\ldots,v_n),
$$

o produto interno usual em $\mathbb{R}^n$ é definido por:

$$
\langle u,v\rangle
=
u_1v_1+u_2v_2+\cdots+u_nv_n
$$

Também podemos escrever:

$$
u \cdot v
$$

ou, na notação matricial,

$$
u^Tv.
$$

O resultado do produto interno entre dois vetores é um **escalar**.

Exemplo

Considere:

$$
u=(1,2,3)
$$

e

$$
v=(4,-1,2).
$$

Então:

$$
u\cdot v
=
1(4)+2(-1)+3(2)
$$

$$
u\cdot v=8.
$$

```python
u = np.array([1, 2, 3])
v = np.array([4, -1, 2])

produto_interno = np.dot(u, v)

print(produto_interno)
```

```python
# Uma maneira mais moderna para realizar:
u @ v
```

# 3.0 Norma de um vetor

A norma representa o comprimento de um vetor.

Se

$$
u=(u_1,u_2,\ldots,u_n),
$$

então sua norma euclidiana é

$$
\|u\|
=
\sqrt{u_1^2+u_2^2+\cdots+u_n^2}.
$$

Como

$$
u\cdot u
=
u_1^2+u_2^2+\cdots+u_n^2,
$$

podemos escrever:

$$
\boxed{
\|u\|=\sqrt{u\cdot u}
}
$$

```python
# Vamos criar um exemplo e fazer a conta na mão:
u = np.array([1, 2, 2])

produto = np.dot(u, u)

norma_manual = np.sqrt(produto)

print(produto)
print(norma_manual)
```

```python
# Agora, usando a biblioteca do Numpy:

norma = np.linalg.norm(u)

print(norma)
```

não é um comando para decorar sem contexto. No caso da norma euclidiana, ele está calculando conceitualmente:

$$ \sqrt{u_1^2+u_2^2+\cdots+u_n^2}. $$

Você pode até comparar os dois:

```python
norma_1 = np.sqrt(np.dot(u, u))
norma_2 = np.linalg.norm(u)

print(norma_1)
print(norma_2)

print(np.isclose(norma_1, norma_2))
```

Interpretação geométrica

Para

$$ u=(3,4), $$

a norma é

$$ \|u\| = \sqrt{3^2+4^2} = 5. $$

Isso é simplesmente Pitágoras:

$$ \sqrt{3^2+4^2}. $$

Em Python:

```python
u = np.array([3, 4])

np.linalg.norm(u)
```

Então, geometricamente, a norma é a distância da origem até o ponto representado pelo vetor.

$$ (0,0) \longrightarrow (3,4) $$

tem comprimento \(5\).


OBS.:  Normalização

O resultado da normalização é um vetor unitário.

Com

$$ u=(6,8), $$

temos

$$ \|u\|=\sqrt{6^2+8^2}=\sqrt{100}=10. $$

Então:

$$ \hat u=\frac{u}{\|u\|} = \frac{(6,8)}{10} = \left(\frac35,\frac45\right). $$

E:

$$ \left\|\left(\frac35,\frac45\right)\right\| = \sqrt{\frac9{25}+\frac{16}{25}} = 1. $$

Ou seja:

$$ \boxed{\text{normalizar um vetor = manter sua direção e transformar seu comprimento em }1} $$

No Python:

```python
u = np.array([6, 8])

u_unitario = u / np.linalg.norm(u)

print(u_unitario)
print(np.linalg.norm(u_unitario))
```

# 4.0 Distância entre vetores

A distância entre dois vetores $u$ e $v$ é dada por:

$$
d(u,v)=\|u-v\|
$$

Ou seja, primeiro calculamos o vetor diferença:

$$
u-v
$$

e depois calculamos sua norma.

Exemplo manual

Considere

$$ u=(1,2) $$

e

$$ v=(4,6). $$

Primeiro:

$$ u-v=(1-4,2-6)=(-3,-4). $$

Agora calculamos a norma:

$$ \|u-v\| = \sqrt{(-3)^2+(-4)^2} $$ $$ = \sqrt{9+16} $$ $$ = \boxed{5}. $$

Portanto:

$$ \boxed{d(u,v)=5}. $$

No Python:

```python
u = np.array([1, 2])
v = np.array([4, 6])

diferenca = u - v

distancia = np.linalg.norm(diferenca)

print("Diferença:", diferenca)
print("Distância:", distancia)
```

```python
# Você também pode fazer diretamente:

distancia = np.linalg.norm(u - v)
distancia
```

A conexão importante

Veja como um conceito está nascendo do outro:

$$ \text{produto interno} \rightarrow \text{norma} \rightarrow \text{distância}. $$

Porque:

$$ \|u\| = \sqrt{u\cdot u} $$

e

$$ d(u,v) = \|u-v\|. $$

Logo, se quisesse escrever tudo em termos de produto interno:

$$ d(u,v) = \sqrt{(u-v)\cdot(u-v)}. $$

Em Python:

```python
distancia_manual = np.sqrt(
    np.dot(u - v, u - v)
)

print(distancia_manual)
```

Interpretação geométrica

Se você pensa em

$$ u=(1,2) $$

e

$$ v=(4,6), $$

como pontos no plano, então a distância entre eles é exatamente a distância euclidiana usual:

$$ \sqrt{(4-1)^2+(6-2)^2}. $$

Ou seja:

$$ \sqrt{3^2+4^2}=5. $$

É a mesma fórmula que você conhece desde geometria analítica. Só estamos escrevendo em linguagem vetorial.

# 5.0 Ângulo entre vetores

O produto interno também permite determinar o ângulo entre dois vetores.

Sabemos que:

$$
u\cdot v
=
\|u\|\|v\|\cos\theta
$$

Logo:

$$
\cos\theta
=
\frac{u\cdot v}{\|u\|\|v\|}
$$

e:

$$
\theta
=
\arccos\left(
\frac{u\cdot v}{\|u\|\|v\|}
\right)
$$

Exemplo manual

Considere:

$$ u=(1,0) $$

e

$$ v=(1,1). $$

Primeiro:

$$ u\cdot v=1. $$

As normas são:

$$ \|u\|=1 $$

e

$$ \|v\|=\sqrt2. $$

Logo,

$$ \cos\theta = \frac{1}{\sqrt2}. $$

Portanto,

$$ \theta = \arccos\left(\frac1{\sqrt2}\right) = \boxed{\frac{\pi}{4}} $$

ou:

$$ \boxed{45^\circ}. $$
Agora em Python

```python
u = np.array([1, 0])
v = np.array([1, 1])

produto = np.dot(u, v)

norma_u = np.linalg.norm(u)
norma_v = np.linalg.norm(v)

cos_theta = produto / (norma_u * norma_v)

theta_rad = np.arccos(cos_theta)

print("Produto interno:", produto)
print("cos(theta):", cos_theta)
print("Ângulo em radianos:", theta_rad)
```

```python
# Caso queira um resultado em graus:
theta_graus = np.degrees(theta_rad)

print(round(theta_graus,2))
```

```python
# para facilitar, podemos criar uma função:

def angulo_entre_vetores(u, v):
    cos_theta = np.dot(u, v) / (
        np.linalg.norm(u) * np.linalg.norm(v)
    )

    theta = np.arccos(cos_theta)

    return np.degrees(theta)
```

```python
u = np.array([1, 0])
v = np.array([1, 1])

print(round(angulo_entre_vetores(u, v),2))
```

# 6.0 Projeção Vetorial

A projeção de um vetor $u$ sobre um vetor $v$ representa a componente de $u$
na direção de $v$.

A fórmula é:

$$
\operatorname{proj}_v(u)
=
\frac{u\cdot v}{v\cdot v}v
$$

Como

$$
v\cdot v = \|v\|^2,
$$

também podemos escrever:

$$
\operatorname{proj}_v(u)
=
\frac{u\cdot v}{\|v\|^2}v
$$

Exemplo manual

Considere:

$$ u=(5,1) $$

e

$$ v=(1,2). $$

Queremos projetar \(u\) sobre \(v\).

Primeiro:

$$ u\cdot v = 5(1)+1(2) = 7. $$

Depois:

$$ v\cdot v = 1^2+2^2 = 5. $$

Então:

$$ \operatorname{proj}_v(u) = \frac75(1,2). $$

Logo:

$$ \boxed{ \operatorname{proj}_v(u) = \left(\frac75,\frac{14}{5}\right) } $$

ou:

$$ \boxed{ \operatorname{proj}_v(u) = (1.4,2.8) } $$



```python
u = np.array([5, 1])
v = np.array([1, 2])

produto_uv = np.dot(u, v)
produto_vv = np.dot(v, v)

projecao = (produto_uv / produto_vv) * v

print("u·v =", produto_uv)
print("v·v =", produto_vv)
print("Projeção =", projecao)
```

```python
# criando a função para projeção:

def projecao(u, v):
    return (np.dot(u, v) / np.dot(v, v)) * v
```

```python
u = np.array([5, 1])
v = np.array([1, 2])

p = projecao(u, v)

print(p)
```
