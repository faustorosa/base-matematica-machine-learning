# 2.0 Representando uma matriz:

Para representar uma matriz, temos duas possibilidades: Arrays ou Matrix.

## 2.1 Array:

Essa é a primeira forma e a mais utilizada para escrever uma matriz. Utilizamos a ideia de vetores e podemos escrever uma matriz como um vetor multidimensional.

```python
m1 = np.array([1,3])
m1
```

Repare que quando escrito com apenas um par de [], temos um vetor, olhe a dimensão:

```python
m1.shape
```

Se quisermos escrever uma matriz linha com duas colunas, precisamos dos dois pares:

```python
m2 = np.array([
    [1,3]
])
m2
```

```python
m2.shape
```

## 2.2 Matriz:

Existe uma outra forma de escrever uma matriz, com a função matrix, porém ela é menos usada. Apenas para deixar registrada:

```python
exemplo_1 = np.matrix([
    [1,2,3],
    [4,5,6],
    [7,8,9]
])
exemplo_1
```

Para verificar a sua dimensão:

```python
exemplo_1.shape
```

# 3.0 Operações com matrizes:

Para começar, vamos criar duas matrizes A e B para podermos ver na prática as operações.

```python
# Criando a matriz A:

A = np.array([
    [1,2,3],
    [4,8,9],
    [0,-1,3.7]
])

# Criando a matriz B:

B = np.array([
    [9,8,-1],
    [0,3,7],
    [7,10,2]
])

# Vamos imprimir as matrizes para vermos:

print(f'''As matrizes A e B são:
{A} 

e 

{B}

, respectivamente''')
```

## 3.1 Soma e subtração:

```python
soma = A + B
sub = A - B

print(f'''
A soma é 
{soma}

A subtração é 
{sub} ''')
```

## 3.2 Multiplicação termo a termo:

Aqui vale ressaltar a diferença entre A*B e A@B, a primeira é a multiplicação termo a termo, enquanto a segunda é realmente o produto matricial.

```python
termo_a_termo = A * B
termo_a_termo
```

Uma outra maneira de obter o mesmo resultado é com a função np.multiply(matriz1,matriz2)

```python
termo_2 = np.multiply(A,B)
termo_2
```

## 3.3 Produto matricial:

```python
produto = A@B
produto
```

Podemos fazer a multiplicação entre matrizes pela dunção np.dot(matriz1, matriz2)

```python
produto_2 = np.dot(A,B)
produto_2
```

## 3.4 Traço de uma matriz:

```python
A
```

```python
traco_A = np.trace(A)

print(f'O traço da matria A é {traco_A}.')
```

# 5.0 Matriz Inversa:

A biblioteca NumPy é realmente bastante completa e, portanto, temos uma função especial para encontrar a matriz inversa. Algo muito útil é que dentro da NumPy temos um pacote de
métodos disponíveis para trabalhar exclusivamente com álgebra linear. Esse pacote se chama (numpy.linalg). Dentro dele podemos encontrar a função para calcular a inversa de uma matriz (chamada de inv()).

```python
inv_A = inv(A)

inv_A
```

# 6.0 Coletando os elementos de uma matriz:

## 6.1 Pegando um único elemento:

Para coletar cada elemento de uma matriz, temos que lembrar que temos uma lista, assim, basta passar o elemento que desejamos. Por exemplo:

```python
# Vamos colocar a matriz A para podermos fazer a verificação

print(A)
```

```python
A11 = A[0,0] # é válido lembrar que começamos com 0 no python
A11
```

## 6.2 Para pegar uma linha inteira:

```python
A_linha_1 = A[0,:]
A_linha_1
```

```python
A_linha_2 = A[1,:]
A_linha_2
```

## 6.3 Para pegar uma coluna inteira:

```python
A_coluna_1 = A[:,0]
A_coluna_1
```

```python
A_coluna_2 = A[:,1]
A_coluna_2
```

OBS.: Para Álgebra Linear, eu prefiro essa segunda forma quando você quer pensar explicitamente em vetores coluna.

```python
A_vetor_coluna_1 = A[:,[0]]
A_vetor_coluna_1
```
