# 1.0 Vetores:

É muito comum termos um conjunto de escalares, um vetor, e precisarmos executar operações de soma, divisão e/ou multiplicação de escalares com esses vetores. Vamos acompanhar o exemplo abaixo:

```python
# Importando a biblioteca Numpy:

import numpy as np
from numpy.linalg import inv
```

```python
# Dado o escalar a:

a = 3

# e o vetor

v = np.array([1,2,3])

print(v)
```

# 2.0 Operações com escalar:

## 2.1 Soma e subtração:

```python
v1 = v + a
v1
```

```python
v2 = v - a
v2
```

## 2.2 Multiplicação:

```python
v3 = a * v
v3
```

## 2.3 Divisão:

```python
v4 = v / a
v4
```

# 3.0 Operações entre vetores:

De maneira análoga, podemos fazer as operações entre vetores também. Assim, temos:

```python
# Já conhecemos o vetor v, vamos criar o vetor  u:

u = np.array([7,8,0])
u
```

Agora, vamos as operações:

```python
soma = v + u
dif = v - u
mult = v * u

soma_2 = 3 * v - 5 * u

print(soma)
print(dif)
print(mult)
print(soma_2)
```
